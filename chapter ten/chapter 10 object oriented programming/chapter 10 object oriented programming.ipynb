## class in programming

```python
class employee:
    name="abhishek"
    language="py"
    salary=15000
abhishek=employee()
print(abhishek.name ,  abhishek.language)    
```

## class attributes

```python
class employee:
    language="py"
    salary=15000
abhishek=employee()
print(abhishek.language ,abhishek.salary )

vineet=employee()
print(vineet.salary,vineet.language)
```

## INSTANCE attributes

```python
class employee:
    language="py"
    salary=15000
abhishek=employee()
abhishek.name="abhishek"
print(abhishek.name,abhishek.language ,abhishek.salary )

vineet=employee()
vineet.name="vineet"
print(vineet.name,vineet.salary,vineet.language)
```

## instance attribute vs class attribute.

```python
class employee:
    language="py"
    name="preeti"
    salary=15000
abhishek=employee()
abhishek.name="abhishek"
print(abhishek.name,abhishek.language ,abhishek.salary )

```

## self parameter.

```python
class employee():
    language="python"
    name="abhishek"
    salary=35000
    def getinfo(self):
        print(f"the salary of abhishek {self.salary}. the language he use in programming {self.language}")
    
abhishek=employee()
abhishek.getinfo()        
```

## static method.

```python
class employee():
    language="python"
    name="abhishek"
    salary=35000
    def getinfo(self):
        print(f"the salary of abhishek {self.salary}. the language he use in programming {self.language}")
    @staticmethod    
    def greet():
        print("good morning")
abhishek=employee()
abhishek.getinfo()        
abhishek.greet()  
```

## dunder method.

```python
class employee():
    language="python"
    name="abhishek"
    salary=35000
    def __init__(self):
        print("this is dunder method:")
    def getinfo(self):
        print(f"the salary of abhishek {self.salary}. the language he use in programming {self.language}")
    @staticmethod    
    def greet():
        print("good morning")
abhishek=employee()
abhishek.getinfo()        
abhishek.greet()  
bali=employee()
```
