## write a class "calculator" capable of finding square, cube and square root of a number.

```python
class calculator:
    def __init__(self,n):
        self.n=n
    @staticmethod
    def greet():
       print("good morning sir")
    def square(self):
        print(f"the square of {self.n*self.n}")
    def cube(self):
        print(f"the square of {self.n*self.n*self.n}")
    def squareroot(self):
        print(f"the square of {self.n**1/2}")
a=calculator(5)
a.square()
a.cube()
a.squareroot()  
a.greet()          


   
```

## write a class  train  which has methods to book  a ticket  , get status (no of seats) and  get fare information of train  running  under indian railway.

```python
from random import randint

class Train:

    def __init__(self, trainNo):
        self.trainNo = trainNo

    def book(self, fro, to):
        print(f"Ticket is booked in train no: {self.trainNo} from {fro} to {to}")

    def getStatus(self):
        print(f"Train no: {self.trainNo} is running on time")

    def getFare(self, fro, to):
        print(f"Ticket fare in train no: {self.trainNo} from {fro} to {to} is {randint(222, 5555)}")


t = Train(12399)

t.book("Rampur", "Delhi")
t.getStatus()
t.getFare("Rampur", "Delhi")
```
