## write a program  to print multiplication table of given number using for loop.

```python
n=int(input("enter number:"))
for i in range(1,11):
    print(f"{n}X{i}={n*i}")
```

## write a program to greet all person names stored in a list 'l' and which starts with s.
## l=["harry","soham","sachin","rahul"]

```python
l=["harry","soham","sachin","rahul"]
for name in l:
    if(name.startswith("s")):
        print(f"hello {name}")
    
```

## attempt problem 1 using while loop.

```python
n=int(input("enter number:"))
i=0
while(i<11):
    print(f"{n}X{i}={n*i}")
    i+=1
```

## write a program   to find whether a given number is prime or not.

```python
n=int(input("enter number:"))

while(n%2==0):
    print("prime")
else:
    print("not prime")
```

## write a program  to find the sum of first n natural numbers using while loop.

```python
n=int(input("enter the number:"))
i=1
sum=0
while(i<=n):
    sum+=i
    i+=1
    print(sum)
```

## write a program to calculate the factorial of given number using for loop.

```python
n=int(input("enter a number:"))
product=1
for i in range(1,n+1):
    product=product*i 
print(f"the factorial of {n} is {product}")    
```

## write a program to print the following star pattern.
##  *
##  ***
## ***** for n=3  

```python
n=int(input("enter the no:"))
for i in range(1,n+1):
    print(" "*(n-1),end="")
    print("*"*(2*i-1),end="")
    print("")
```

## write a program to print multiplicaiton table of n using for loop in reversed order.

```python
n=int(input("enter the number:"))
for i in range(1,11):
    print(f"{n}X{11-i}={n*(11-i)}")
```
