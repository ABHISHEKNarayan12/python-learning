## quick quiz write a program  to print the  content of a list using while loops.

```python
fruits = ["Apple", "Banana", "Mango", "Orange", "Grapes", "Pineapple", "Watermelon", "Papaya"]
i=0
while(i<len(fruits)):
    print(fruits[i])
    i+=1
```

## for loop iterate.


```python
# list
l=[34,2,76,35,74,26,99]
for i in l:
    print(i)
l1=(34,2,76,35,74,26,99)
for i in l1:
    print(i)
l2="34,2,76,35,74,26,99"
for i in l2:
    print(i)
```

## for loops with else.

```python
l=[1,3,5,7,9]
for i in l:
    print(i)
else:
    print("done bro")
```

## the break statement

```python
for i in range(100):
    if(i==10):
        break
    print(i)
```

## the continue statement

```python
for i in range(100):
    if(i==10):
        continue
    print(i)
```
