## write a program using functions to find greatest of three numbers. 


```python
def gotn():
    n1=int(input("enter the no 1 :"))
    n2=int(input("enter the no 2 :"))
    n3=int(input("enter the no 3 :"))
    if(n1>n2 and n1>n3):
        print(f"n1 is greatest :{n1}")
    elif(n2>n1 and n2>n3):
         print(f"n2 is greatest :{n2}")
    else:
        print(f"n3 is greatest :{n3}")
gotn()        
```

## write a python program using function to convert  celcius to fahrenheit.

```python
def convert():
    celcius=int(input("enter the value for celcius:"))    
    fahrenheit = (celcius * 9/5) + 32
    print(fahrenheit)
convert()    
```

## write a recursive function to calculate the sum of first n natural numbers.

```python
def fsum(n):
    if(n==1):
        return 1
    return fsum(n-1)+n  
print(fsum(5))    
```

## write a python function to print first n lines of the following pattern.
## ***
## **    for n=3
## *

```python
def pattern(n):
    if(n==0):
        return
    print("*"*n)
    pattern(n-1)
pattern(3)
```

## write a python function which  converts inches to cms.

```python
def inch_to_cms(inch):
    return inch *2.54
n=int(input("enter values in inches:"))
print(f"the corresponding value {n} in cms is {inch_to_cms(n)}")
```

## write a python function to remove a given word from a list ad strip it at the  same time. 

```python
def rem(l,word):
    n=[]
    for item in l:
        if not(item==word):
            n.append(item.strip(word))
    return n
l=["harry","rohan","abhishek","vivek","amit"]
print(rem(l,"a"))            
```

## write a python function to print multiplication table of a given number.

```python
def table(n):
    for i in range(1,11):
        print(f"{n}X{i}={n*i}")

table(4)        
    
```
