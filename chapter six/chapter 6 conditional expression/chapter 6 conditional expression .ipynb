##  If & Else conditional 

```python
age=int(input("enter your age: "))
if(age>=18):
    print("your age above the age of 18 so you  gain the driving licence:")
else:
    print("your age is not above the 18 so you  not enough to gain the driving licence:")
```

## if elif & else ladder.

```python
age=int(input("enter your age: "))
if(age>=18):
    print("your age above the age of 18 so you  gain the driving licence:")
elif(age<0):
    print("you are entering the invalid negative age age:")
elif(age==0):
    print("you are entering the  age 0 which is invalid:")

else:
    print("your age is not above the 18 so you  not enough to gain the driving licence:")
```
