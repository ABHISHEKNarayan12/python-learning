## write a program  to find the greatest of four numbers entered by the user.

```python
n1=int(input("enter the no 1:"))
n2=int(input("enter the no 2:"))
n3=int(input("enter the no 3:"))
n4=int(input("enter the no 4:"))
if(n1>n2 and n1>n3 and n1>n4):
   print("greater number n1 is",n1)
elif(n2>n1 and n2>n3 and n2>n4):
   print("greater number n1 is",n2)
elif(n3>n1 and n3>n2 and n3>n4):
   print("greater number n1 is",n3)
else:
    print("n4 is greatest no:",n4)
```

## write a program  to find out whether a student has passed or failed if it requires a total of 40% and at least 33% in each subject to pass . assume 3 subject and take marks as an input from user.

```python
maths=int(input("enter the marks of maths:"))
english=int(input("enter the marks of english:"))
science=int(input("enter the marks of science:"))
total_percentage=(100)*(maths+english+science)/300
if(total_percentage>=40):
   print("your marks is above 33% your  pass in exam :",total_percentage)
else:
   print("your marks is  33% your are failed in exam but the conditions make you  pass in exam :")
```

## a spam comment is defined as a text containing following keywords:
## make a lot of money , buy now , subscribe this , click here , write a program to detect these spams

##  write a program to calculate the grade of a student  from his marks from  the  following scheme:
## 90-100=>ex
## 80-90=>a
## 75-85=>b
## 60-75=>c
## 50-60=>d
## 50=>f

```python
marks=int(input("enter your marks:"))
if(marks<=100 and marks>=90):
    grade="ex"
elif(marks<=90 and marks>=80):
    grade="A"
elif(marks<=80 and marks<=75):
    grade="B"
elif(marks<=70 and marks>=60):
    grade="C"
elif(marks<=60 and marks>=50):
    grade="D"
elif(marks<=50 ):
    grade="F"
print("your grade is :",grade)
```
