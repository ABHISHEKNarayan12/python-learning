## item assigment in list 

```python
things=[1,True,False , 34.5, "abhishek"]
things[0]="2"
print(things[0])
```

## slicing in list

```python
things=[1,True,False , 34.5, "abhishek"]
things[0]="2"
print(things[0:])
```
