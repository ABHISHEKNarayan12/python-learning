## dictionary methods .

```python
studentmarks={
    "abhishek":70,
    "shubham":63,
    "muski":99,
    "list":[1,2,3,4]
}
print(studentmarks.items())
print(studentmarks.keys())
print(studentmarks.values())
studentmarks.update({"abhishek":90,"muski":89})
print(studentmarks)
print(studentmarks.get("vinay"))
print(studentmarks.get("muski1")) 
print(studentmarks["muski"])
```
