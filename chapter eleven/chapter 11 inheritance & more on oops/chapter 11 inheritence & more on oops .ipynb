## multilevel inheritance.

```python
class employee:
     company="ITC"
     name="abhishek"
     salary=30000
     def show(self):
         print(f"the programmer company name is {self.company}\n the programmer name is {self.name}\n the salary of programmer is {self.salary}\n")
   
class coder:
    language="python"
    age=21
class buisness:
    revenue="5B$"
   

class programmer(employee,coder,buisness):
    company="ITC-INFOTECH"
    def showlanguage(self):
        print(f"the name of progammer  {self.name}\n the salary of progammer  {self.salary}\n the age of progammer  {self.age}\n the language use the programmer is {self.language}\n company total revenue {self.revenue} \n")


a=employee()
b= programmer()
c=coder()
b.show()





class low:
    def __init__(self):
     print("the constructor  of low")

    a=5
class mid(low):
    def __init__(self):
     print("the constructor  of mid")
    b=4
class high(mid):
    def __init__(self):
     super().__init__()
     print("the constructor  of high")
    c=5
o=high()
print(o.a, o.b, o.c)
  
```

## class method.

```python
class employee:
    name="abhishek"
    @classmethod
    def show(cls):
        print(f"the name of employee is {cls.name}")
   
e=employee()

e.name="abhinav"
e.show()        
        
```

## property decorator.

```python
class employee:
    name="abhishek"
    @classmethod
    def show(cls):
        print(f"the name of employee is {cls.name}")
    @property
    def name(self):
        return f"{self.fname} {self.lname}"    
    @name.setter
    def name (self,value):
        self.fname=value.split(" ")[0]
        self.lname=value.split(" ")[1]
e=employee()
e.name="abhishek narayan"
print(e.fname ,e.lname)
e.show()        
        
```

## operator overloading in python.

```python
class number:
    def __init__(self ,n):
        self.n=n
    def __add__(self ,num):
        return self.n+num.n
    def __sub__(self ,num):
        return self.n-num.n
    def __mul__(self ,num):
        return self.n*num.n
    def __truediv__(self ,num):
        return self.n/num.n
    def __floordiv__(self ,num):
        return self.n//num.n
n=number(4)
m=number(12)
print(n+m)
print(n-m)
print(n*m)
print(n/m)
print(n//m)
```
