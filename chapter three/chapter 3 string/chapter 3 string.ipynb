## string

```python
name="abhishek"
nameshort=len(name)
print(nameshort)
print(name[0:3])
print(name[1:5])
print(name[:5])
print(name[1:])
```

## negative slicng value

```python
name="abhishek"
print(name[-8:-1])
```

## string function

```python
name="abhishek"
print(len(name))
print(name.endswith("k"))
print(name.startswith("a"))
print(name.count("s"))
print(name.capitalize())
print(name.find("i"))
print(name.replace("shek","jeet"))
```
