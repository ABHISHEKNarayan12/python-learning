## the game() function in a program lets a user play a game and return the score as an integer. you need to read a file 'hi-score.txt' which is either blank or contains the previous hi-score. you need to write a program to update the hi-score whenever the game() function breaks the hi-score.

```python
import random 
def game():
    print("your are playing game...")
    score=  random.randint(1,100)
    with open("hiscore.txt")as f:
        hiscore=f.read()
        if(hiscore!=""):
            hiscore=int(hiscore)
        else:
            hiscore=0
    print(f"your score:{score}")
    if(score>hiscore):
        with open("hiscore.txt","w") as f:
            f.write(str(score))

        return score
game()                

        
```

## write a program to generate a multiplication tables from 2 to 20 and write it to the different files .place these files in folder for a 13 -years old.

```python
def genratetable(n):
    table=""
    for i in range(1,11):
        table+=f"{n}X{i}={n*i}\n"
    with open(f"tables/table_{n}","w")as f:
        f.write(table)
    
for i in range(2,21) :
    genratetable(i)       
```

## repeat program 4 for a list of such words to be censored.

```python
words=["kamjor","ill","donkey","purpose","weak"]
with open("animal.txt", "r")as f:
    content=f.read()
    for word in words:
        contentnew=content.replace(word,"#"*len(word))
with open("animal.txt","w")as f:
    f.write(contentnew)
```
